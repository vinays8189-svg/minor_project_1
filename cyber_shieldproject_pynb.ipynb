password len-checker

```python
import re

def analyze_password(password):
    checks = {
        "At least 12 characters": len(password) >= 12,
        "Has lowercase letter": bool(re.search(r"[a-z]", password)),
        "Has uppercase letter": bool(re.search(r"[A-Z]", password)),
        "Has a number": bool(re.search(r"\d", password)),
        "Has special character": bool(re.search(r"[^A-Za-z0-9]", password)),
        "No repeated character 3+ times": not bool(re.search(r"(.)\1\1", password)),
        "Not a common password": password.lower() not in [
            "password", "password123", "123456", "qwerty", "admin"
        ]
    }

    score = round(sum(checks.values()) / len(checks) * 100)

    return {
        "score": score,
        "checks": checks,
        "recommendation": (
            "Use a unique password with 12+ characters, numbers, and symbols."
            if score < 80 else
            "Strong password. Keep it unique and enable MFA."
        )
    }

# Sample test — do not use your real password in a presentation
password_result = analyze_password("DemoPass!2026")
print(password_result)
```

suspicious url checker

```python
def check_url(url):
    if not url.startswith(("http://", "https://")):
        url = "https://" + url

    parsed = urlparse(url)
    host = parsed.hostname or ""
    flags = []

    suspicious_tlds = ["zip", "top", "xyz", "click", "tk"]
    suspicious_words = ["login", "verify", "secure", "update", "password", "account"]

    if parsed.scheme != "https":
        flags.append("Website does not use HTTPS")

    if "@" in parsed.netloc:
        flags.append("URL contains @ and may hide the true website")

    if re.fullmatch(r"\d{1,3}(\.\d{1,3}){3}", host):
        flags.append("Uses an IP address instead of a domain")

    if host.count(".") >= 3:
        flags.append("Contains many subdomains")

    if "xn--" in host:
        flags.append("Uses punycode and may imitate another website")

    if len(url) > 100:
        flags.append("URL is unusually long")

    if any(host.endswith("." + tld) for tld in suspicious_tlds):
        flags.append("Uses a potentially high-risk domain extension")

    if any(word in url.lower() for word in suspicious_words):
        flags.append("Contains account or credential-related words")

    risk_points = min(30, len(flags) * 4)

    if risk_points < 8:
        verdict = "Low"
    elif risk_points < 18:
        verdict = "Medium"
    else:
        verdict = "High"

    return {
        "url": url,
        "flags": flags,
        "risk_points": risk_points,
        "verdict": verdict
    }

url_result = check_url("http://verify-account.top/login")
print(url_result)
```

network port rish checker

```python
def check_local_ports():
    risky_ports = {
        21: "FTP",
        23: "Telnet",
        80: "HTTP",
        135: "Windows RPC",
        139: "NetBIOS",
        445: "SMB",
        3389: "Remote Desktop Protocol"
    }

    open_ports = []
    risk_points = 0

    for port, service in risky_ports.items():
        sock = socket.socket(socket.AF_INET, socket.SOCK_STREAM)
        sock.settimeout(0.3)

        result = sock.connect_ex(("127.0.0.1", port))
        sock.close()

        if result == 0:
            open_ports.append({
                "port": port,
                "service": service
            })
            risk_points += 8

    return {
        "open_ports": open_ports,
        "risk_points": min(risk_points, 35)
    }

# Use only on your own computer, not needed for the Colab demo:
# network_result = check_local_ports()
# print(network_result)
```

```python
import re
from urllib.parse import urlparse

def analyze_password(password):
    checks = {
        "length": len(password) >= 12,
        "lowercase": bool(re.search(r"[a-z]", password)),
        "uppercase": bool(re.search(r"[A-Z]", password)),
        "number": bool(re.search(r"\d", password)),
        "special_character": bool(re.search(r"[^A-Za-z0-9]", password))
    }

    score = round(sum(checks.values()) / len(checks) * 100)

    return {
        "score": score,
        "recommendation": (
            "Use a unique password with 12+ characters, numbers, and symbols."
            if score < 80 else "Strong password."
        )
    }

def check_url(url):
    if not url.startswith(("http://", "https://")):
        url = "https://" + url

    parsed = urlparse(url)
    host = parsed.hostname or ""
    flags = []

    if parsed.scheme != "https":
        flags.append("Website does not use HTTPS")

    if any(word in url.lower() for word in ["login", "verify", "password", "account"]):
        flags.append("Contains account or credential-related wording")

    if host.endswith((".top", ".xyz", ".zip", ".tk")):
        flags.append("Potentially high-risk domain extension")

    points = min(30, len(flags) * 6)
    verdict = "Low" if points < 8 else "Medium" if points < 18 else "High"

    return {
        "flags": flags,
        "risk_points": points,
        "verdict": verdict
    }

def detect_phishing(message):
    phrases = ["urgent", "immediately", "click here", "suspended", "verify", "password"]
    flags = []

    for phrase in phrases:
        if phrase in message.lower():
            flags.append(f"Suspicious phrase: {phrase}")

    if message.count("!") >= 3:
        flags.append("Excessive urgency symbols")

    points = min(30, len(flags) * 5)
    verdict = "Low" if points < 8 else "Medium" if points < 18 else "High"

    return {
        "flags": flags,
        "risk_points": points,
        "verdict": verdict
    }

def generate_cyber_score(password, url, message):
    password_result = analyze_password(password)
    url_result = check_url(url)
    phishing_result = detect_phishing(message)

    password_risk = round((100 - password_result["score"]) * 0.25)

    total_risk = (
        password_risk
        + url_result["risk_points"]
        + phishing_result["risk_points"]
    )

    security_score = max(0, min(100, 100 - total_risk))

    if security_score >= 80:
        risk_level = "Low Risk"
    elif security_score >= 60:
        risk_level = "Moderate Risk"
    elif security_score >= 40:
        risk_level = "High Risk"
    else:
        risk_level = "Critical Risk"

    return {
        "security_score": security_score,
        "risk_level": risk_level,
        "password_result": password_result,
        "url_result": url_result,
        "phishing_result": phishing_result
    }

result = generate_cyber_score(
    password="DemoPass!2026",
    url="http://verify-account.top/login",
    message="URGENT!!! Your account is suspended. Click here immediately to verify your password!"
)

print("CyberShield is ready.")
print("Security Score:", result["security_score"], "/100")
print("Risk Level:", result["risk_level"])
```

```python
power_bi_rows = []

for flag in result["url_result"]["flags"]:
    power_bi_rows.append([
        "URL Security",
        result["url_result"]["verdict"],
        result["url_result"]["risk_points"],
        flag
    ])

for flag in result["phishing_result"]["flags"]:
    power_bi_rows.append([
        "Phishing Detection",
        result["phishing_result"]["verdict"],
        result["phishing_result"]["risk_points"],
        flag
    ])

power_bi_rows.append([
    "Password Security",
    "Low" if result["password_result"]["score"] >= 80 else "Medium",
    100 - result["password_result"]["score"],
    result["password_result"]["recommendation"]
])

with open("cybershield_powerbi.csv", "w", newline="", encoding="utf-8") as file:
    writer = csv.writer(file)
    writer.writerow(["Category", "Risk_Level", "Risk_Points", "Finding"])
    writer.writerows(power_bi_rows)

print("Power BI file created successfully!")
```

```python
import hashlib

# Create a safe sample file for the demonstration
file_name = "sample_file.txt"

with open(file_name, "w", encoding="utf-8") as file:
    file.write("This is a safe sample file created for the CyberShield project.")

# Calculate SHA-256 hash
sha256 = hashlib.sha256()

with open(file_name, "rb") as file:
    for chunk in iter(lambda: file.read(4096), b""):
        sha256.update(chunk)

hash_result = {
    "file_name": file_name,
    "sha256_hash": sha256.hexdigest(),
    "status": "File hash generated successfully"
}

print(hash_result)
```
